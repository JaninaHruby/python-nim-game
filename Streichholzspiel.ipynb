Baue ein Barspiel bei dem eine zufällige anzahl an Streichhölzern zwischen 21 und 29
verwendet wird. Jeder Spieler muss zwischen 1-3 Streichhölzer wegnehmen pro Runde. Der, der das letzte Streichholz nimmt verliert

```python
import numpy as np
import random
from PIL import Image
import matplotlib.pyplot as plt
import math
import os

def streichholz_anzeigen(anzahl):
    # --- 1. Configuration ---
    dateiname = "streichholz.png"
    ziel_groesse = 50 
    
    # --- 2. Grid calculation (Single row) ---
    zeilen = 1 
    spalten = math.ceil(anzahl / zeilen) 

    # --- 3. Load and prepare image ---
    try:
        streichholz_objekt = Image.open(dateiname)
        # NEW: Resize image even smaller (e.g., 30x30) to save space
        streichholz_klein = streichholz_objekt.resize((30, 30)) 
        bild_array = np.array(streichholz_klein)
        
    except FileNotFoundError:
        print(f"ERROR: Image '{dateiname}' not found in path: {os.getcwd()}")
        return 

    # --- 4. Create and fill subplots ---
    # figsize is scaled to a very small base (0.3 inches per image)
    fig, axes = plt.subplots(zeilen, spalten, figsize=(spalten * 0.3, zeilen * 0.3)) 
    
    # Simplify axis structure
    if zeilen * spalten > 1:
        axes = axes.flatten()
    else:
        axes = [axes]

    # Fill subplots
    for i in range(anzahl):
        ax = axes[i]
        ax.imshow(bild_array)
        ax.axis('off')

    # --- 5. Remove unused axes ---
    for i in range(anzahl, len(axes)):
        fig.delaxes(axes[i])

    # --- 6. Rendering (NEW: Aggressive margin adjustment) ---
    # Sets all margins (left, right, top, bottom) to near zero
    # to keep the window as small as possible
    plt.subplots_adjust(left=0.01, right=0.99, top=0.99, bottom=0.01)

    plt.show()

# --- Example function call (optional) ---
# streichholz_anzeigen(anzahl=30)

# Define the range of matches and output a random number of matches for the game
anzahl = range (21, 30)
spiel = (random.choice(anzahl))
streichholz_anzeigen(spiel)
print(f"({spiel})")

while spiel > 0:
    
    # Ask the player how many matches they want to take. Handle non-integer inputs.
    try:
        take = int(input("wieviele nimmst du? (1-3 möglich)"))
    except ValueError:
        print("Das ist keine gültige Zahl, du verlierst!")
        break
        
    # Check if the player follows the rules; otherwise, they lose
    if take > 3 or take <= 0: 
        print("Spiel ordentlich! Du hast veroren!")
        break

    # If only 1 match remains before my move, I lose
    if spiel == 1:
        print("Du hast verloren")
        break
        
    # Remove the chosen number of matches
    spiel -= take
    streichholz_anzeigen(spiel)

    # If only 1 remains before their move, they lose
    if spiel == 1:
        print("PC hat verloren")
        break
        
    # Computer makes its move using the implemented winning strategy 

    if spiel % 4 == 0: zug = 3 
    
    if spiel % 4 == 1: zug = 1

    if spiel % 4 == 2: zug = 1
    
    if spiel % 4 == 3: zug = 2

    print(f"Ich wähle {zug}")
    spiel -= zug
    streichholz_anzeigen(spiel)
    print(f"({spiel})")
```
